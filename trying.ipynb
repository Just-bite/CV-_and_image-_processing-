```python
from matplotlib.figure import Figure
x_array= [i for i in range(0, 10)]
y_array= [x_array[j]**2 for j  in range(0, 10)]
figure,axes = plt.subplots(1,3,figsize = (10,5))

axes[0].set_title("Kore")
axes[0].set_xlim(0,10)
axes[0].set_ylim(0,81)
axes[0].plot(x_array, y_array)

axes[1].set_title("Youang")
axes[1].set_xlim(0,10)
axes[1].set_ylim(0,10)

axes[2].set_title("Widow empress")
axes[1].set_xlim(0,10)
axes[1].set_ylim(0,10)

plt.show()
```

```python
def f(t):
    return np.exp(t) * np.cos(2*np.pi*t)

t1 = np.arange(0.0, 5.0, 0.1)
t2 = np.arange(0.0, 5.0, 0.02)

plt.figure()
plt.subplot(211)
plt.plot(t1, f(t1), 'bo', t2, f(t2), 'k')

plt.subplot(212)
plt.plot(t2, np.cos(2*np.pi*t2), 'r--')
plt.show()
```

```python
import cv2 as cv
```

```python
figure, axes = plt.subplots(1, 2, figsize=(10,5))
axes[0].axis('off')
axes[0].set_title("before")
axes[1].axis('off')
axes[1].set_title("after")

pixels = cv.imread("test.jpg", cv.IMREAD_GRAYSCALE)

if pixels is not None:
  axes[0].imshow(pixels, cmap= 'gray')
  pixels= cv.bitwise_not(pixels)

axes[1].imshow(pixels, cmap= 'gray')

plt.show()
```

```python
figure,axes = plt.subplots(1, 2, figsize=(10,5))

axes[0].set_title("source image")
axes[1].set_title("increased brightness by 50")

axes[0].axis('off')
axes[1].axis('off')

source_image_pixels = cv.imread("test.jpg", 0)
result_image_pixels = source_image_pixels.copy()

axes[0].imshow(source_image_pixels, cmap= 'gray')

rows, cols = source_image_pixels.shape
amplifier: int = 50

for row in range(rows):
  for col in range(cols):

    if source_image_pixels[row, col] > 255 - amplifier:
      result_image_pixels[row, col] = 255
    else:
      result_image_pixels[row, col] = source_image_pixels[row, col] + amplifier

axes[1].imshow(result_image_pixels, cmap ='gray')
```

```python
figure, axes = plt.subplots(1,2, figsize=(10,5))

axes[0].set_title("source image")
axes[1].set_title("rotated 90 clockwise")

axes[0].axis('off')
axes[1].axis('off')

source_image_pixels_bgr = cv.imread("drive/MyDrive/Download/01.jpg")
source_image_pixels_rgb = cv.cvtColor(source_image_pixels_bgr, cv.COLOR_BGR2RGB)

axes[0].imshow(source_image_pixels_rgb)

rows,cols,channels = source_image_pixels_bgr.shape

center = (cols / 2, rows / 2)
angle = -90

rad = np.deg2rad(angle)

new_height = int(abs(rows * np.cos(rad) + cols * np.sin(rad)))
new_width = int(abs(cols * np.cos(rad) + rows * np.sin(rad)))

print(cols, rows)
print(new_width, new_height)

M = cv.getRotationMatrix2D(center, angle, 1)
M[0, 2] += (rows - new_height) / 2  # коррекция по X
M[1, 2] += (cols - new_width) / 2  # коррекция по Y

result_image_pixels_bgr = cv.warpAffine(source_image_pixels_bgr, M, (new_width, new_height))
result_image_pixels_rgb = cv.cvtColor(result_image_pixels_bgr, cv.COLOR_BGR2RGB)

axes[1].imshow(result_image_pixels_rgb)
```
