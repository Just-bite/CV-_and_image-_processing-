```python
figure,axes = plt.subplots(1, 2, figsize=(10,5))

axes[0].set_title("source image")
axes[1].set_title("increased brightness by 50")

axes[0].axis('off')
axes[1].axis('off')

source_image_pixels = cv.imread("test.jpg", 0)
result_image_pixels = source_image_pixels.copy()

axes[0].imshow(source_image_pixels, cmap= 'gray')

rows, cols = source_image_pixels.shape
amplifier = 50

for row in range(rows):
  for col in range(cols):

    if source_image_pixels[row, col] > 255 - amplifier:
      result_image_pixels[row, col] = 255
    else:
      result_image_pixels[row, col] = source_image_pixels[row, col] + amplifier

axes[1].imshow(result_image_pixels, cmap ='gray')
```

```python
figure, axes = plt.subplots(1,2, figsize=(10,5))

axes[0].set_title("source image")
axes[1].set_title("rotated")

axes[0].axis('off')
axes[1].axis('off')

source_image_pixels_bgr = cv.imread("drive/MyDrive/Download/01.jpg")
source_image_pixels_rgb = cv.cvtColor(source_image_pixels_bgr, cv.COLOR_BGR2RGB)

axes[0].imshow(source_image_pixels_rgb)

rows,cols,channels = source_image_pixels_bgr.shape

center = (cols / 2, rows / 2)

angle = int(input())
rad = np.deg2rad(angle)

print(f"rotating on {angle}")

new_height = int(abs(rows * np.cos(rad)) + abs(cols * np.sin(rad)))
new_width = int(abs(cols * np.cos(rad)) + abs(rows * np.sin(rad)))

M = cv.getRotationMatrix2D(center, angle, 1)

center_x = cols / 2
center_y = rows / 2

rotated_center_x = M[0, 0] * center_x + M[0, 1] * center_y + M[0, 2]
rotated_center_y = M[1, 0] * center_x + M[1, 1] * center_y + M[1, 2]

new_center_x = new_width / 2
new_center_y = new_height / 2

M[0, 2] += new_center_x - rotated_center_x
M[1, 2] += new_center_y - rotated_center_y

result_image_pixels_bgr = cv.warpAffine(source_image_pixels_bgr, M, (new_width, new_height))
result_image_pixels_rgb = cv.cvtColor(result_image_pixels_bgr, cv.COLOR_BGR2RGB)

axes[1].imshow(result_image_pixels_rgb)

plt.tight_layout()
plt.show()
```

```python
figure, axes = plt.subplots(1,4,figsize=(15,5))

axes[0].set_title("source image")
axes[1].set_title("simple border detector horizontal")
axes[2].set_title("simple border detector vertical")
axes[3].set_title("simple combination of both detectors")

axes[0].axis('off')
axes[1].axis('off')
axes[2].axis('off')
axes[3].axis('off')

source_image = cv.imread("drive/MyDrive/Download/01.jpg", 0)
result_image_v = source_image.copy()
result_image_h = source_image.copy()
result_image = source_image.copy()

axes[0].imshow(source_image, cmap= 'gray')

rows, cols = source_image.shape

for row in range(rows):
  for col in range(cols - 1):
    if(source_image[row, col + 1] <= source_image[row, col]):
      result_image_v[row, col] -= source_image[row, col + 1]
    else:
      result_image_v[row, col] = 0

axes[1].imshow(result_image_v, cmap= 'gray')

for col in range(cols):
  for row in range(rows - 1):
    if(source_image[row + 1, col] <= source_image[row, col]):
      result_image_h[row, col] -= source_image[row + 1, col]
    else:
      result_image_h[row, col] = 0

axes[2].imshow(result_image_h, cmap= 'gray')

for row in range(rows):
  for col in range(cols):
    result_image[row, col] = max(result_image_v[row, col], result_image_h[row, col])

axes[3].imshow(result_image, cmap= 'gray')

plt.tight_layout()
plt.show()
```

```python
def get_filtered_value(source_image, row, col):
  kernel =np.array([[-1, -1, -1],
                    [-1,  9, -1],
                    [-1, -1, -1]])
  new_value = 0
  kernel_rows,kernel_cols = kernel.shape
  for i in range(kernel_rows):
    for j in range(kernel_cols):
      new_value += source_image[row + i -1, col + j - 1] * kernel[i,j]

  return np.clip(new_value, 0, 255)


figure,axes = plt.subplots(1,2,figsize=(10,5))

axes[0].set_title("source image")
axes[1].set_title("sharpening mine")

axes[0].axis('off')
axes[1].axis('off')

source_image = cv.imread("drive/MyDrive/Download/01.jpg", 0)
result_image = source_image.copy().astype(float)

axes[0].imshow(source_image, cmap='gray')

rows,cols = source_image.shape

for row in range(1, rows - 1):
  for col in range(1, cols - 1):
    result_image[row,col] = get_filtered_value(source_image, row, col)

result_image = result_image.astype(np.uint8)

axes[1].imshow(result_image, cmap ='gray')

plt.tight_layout()
plt.show()
```
